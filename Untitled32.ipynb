```python
import fitz
import pandas as pd
import re
import sqlite3
from google.colab import files


# ============================================================
# FUNÇÕES AUXILIARES
# ============================================================

def numero_br(valor):
    """
    Converte:
    20.790   -> 20790
    2,50     -> 2.50
    51.975,00 -> 51975.00
    """
    valor = str(valor).strip()
    valor = valor.replace(".", "").replace(",", ".")
    return float(valor)


meses = {
    "Janeiro": 1,
    "Fevereiro": 2,
    "Março": 3,
    "Abril": 4,
    "Maio": 5,
    "Junho": 6,
    "Julho": 7,
    "Agosto": 8,
    "Setembro": 9,
    "Outubro": 10,
    "Novembro": 11,
    "Dezembro": 12,
}


# ============================================================
# ABRIR PDF
# ============================================================

doc = fitz.open(PDF_PATH)

registros = []

mes_atual = None
ano_atual = None

texto_completo = ""


# ============================================================
# LER PÁGINA POR PÁGINA
# ============================================================

for pagina_num, pagina in enumerate(doc, start=1):

    texto = pagina.get_text("text")
    texto_completo += "\n" + texto

    linhas = texto.splitlines()

    for linha in linhas:

        linha = linha.strip()

        if not linha:
            continue

        # ----------------------------------------------------
        # IDENTIFICAR MÊS
        # Exemplo: Janeiro 2025
        # ----------------------------------------------------

        match_mes = re.match(
            r"^(Janeiro|Fevereiro|Março|Abril|Maio|Junho|Julho|Agosto|"
            r"Setembro|Outubro|Novembro|Dezembro)\s+(\d{4})$",
            linha,
            flags=re.IGNORECASE
        )

        if match_mes:

            nome_mes_encontrado = match_mes.group(1)

            # Corrige capitalização
            for nome in meses:
                if nome.lower() == nome_mes_encontrado.lower():
                    mes_atual = nome
                    break

            ano_atual = int(match_mes.group(2))

            continue


        # ----------------------------------------------------
        # IDENTIFICAR VENDA
        #
        # Exemplo:
        # PET - 20.790 kg - R$ 2,50/kg - Total R$ 51.975,00
        # ----------------------------------------------------

        match_venda = re.match(
            r"^(.+?)\s*-\s*"
            r"([\d.,]+)\s*kg\s*-\s*"
            r"R\$\s*([\d.,]+)\s*/kg\s*-\s*"
            r"Total\s*R\$\s*([\d.,]+)",
            linha,
            flags=re.IGNORECASE
        )

        if match_venda and mes_atual:

            material = match_venda.group(1).strip()

            quantidade = numero_br(
                match_venda.group(2)
            )

            preco = numero_br(
                match_venda.group(3)
            )

            total = numero_br(
                match_venda.group(4)
            )

            registros.append({
                "ano": ano_atual,
                "mes_numero": meses[mes_atual],
                "mes": mes_atual,
                "material": material,
                "quantidade_kg": quantidade,
                "preco_kg": preco,
                "valor_total": total,
                "pagina": pagina_num
            })


doc.close()


# ============================================================
# CRIAR BANCO DE DADOS
# ============================================================

df = pd.DataFrame(registros)

df.insert(
    0,
    "id_venda",
    range(1, len(df) + 1)
)


# ============================================================
# CHECK FINANCEIRO
# ============================================================

df["valor_calculado"] = (
    df["quantidade_kg"]
    *
    df["preco_kg"]
).round(2)

df["diferenca"] = (
    df["valor_total"]
    -
    df["valor_calculado"]
).round(2)

df["valor_ok"] = (
    df["diferenca"].abs() < 0.01
)


# ============================================================
# PEGAR FATURAMENTO TOTAL INFORMADO NO PDF
# ============================================================

match_faturamento = re.search(
    r"Faturamento\s+Total\s+Estimado\s+de\s+2025:\s*R\$\s*([\d.,]+)",
    texto_completo,
    flags=re.IGNORECASE | re.DOTALL
)

if match_faturamento:
    faturamento_pdf = numero_br(
        match_faturamento.group(1)
    )
else:
    faturamento_pdf = None


faturamento_extraido = df["valor_total"].sum()


print("Quantidade de registros:", len(df))
print()
print("Soma das vendas extraídas: R$",
      f"{faturamento_extraido:,.2f}")

print("Faturamento informado no PDF: R$",
      f"{faturamento_pdf:,.2f}"
      if faturamento_pdf is not None
      else "não identificado")

if faturamento_pdf is not None:
    print(
        "Diferença: R$",
        f"{faturamento_pdf - faturamento_extraido:,.2f}"
    )


display(df)


# ============================================================
# RESUMO POR MÊS
# ============================================================

resumo_mensal = (
    df
    .groupby(
        ["ano", "mes_numero", "mes"],
        as_index=False
    )
    .agg(
        quantidade_kg=("quantidade_kg", "sum"),
        faturamento=("valor_total", "sum")
    )
    .sort_values(
        ["ano", "mes_numero"]
    )
)

display(resumo_mensal)


# ============================================================
# SALVAR
# ============================================================

ARQUIVO_EXCEL = "Base_Vendas_Reciclagem_2025.xlsx"
ARQUIVO_CSV = "Base_Vendas_Reciclagem_2025.csv"
ARQUIVO_DB = "Base_Vendas_Reciclagem_2025.db"


# Excel
with pd.ExcelWriter(
    ARQUIVO_EXCEL,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="vendas",
        index=False
    )

    resumo_mensal.to_excel(
        writer,
        sheet_name="resumo_mensal",
        index=False
    )


# CSV
df.to_csv(
    ARQUIVO_CSV,
    index=False,
    sep=";",
    decimal=",",
    encoding="utf-8-sig"
)


# SQLite
con = sqlite3.connect(ARQUIVO_DB)

df.to_sql(
    "vendas",
    con,
    if_exists="replace",
    index=False
)

con.close()


print("\nArquivos gerados:")
print(ARQUIVO_EXCEL)
print(ARQUIVO_CSV)
print(ARQUIVO_DB)


# Baixa o Excel automaticamente
files.download(ARQUIVO_EXCEL)
```
