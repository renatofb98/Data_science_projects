```python
import pandas as pd

# Load the Excel file
file_path = '/content/Simulado 1º dia Inglês_2025-09-05.xlsx'
df = pd.read_excel(file_path)

# Strip whitespace from column names
df.columns = df.columns.str.strip()

# Identify the id columns (like 'MATRÍCULA') and the question columns
id_vars = ['MATRÍCULA'] # Add other identifying columns if necessary
question_cols = [col for col in df.columns if col.startswith('Q ')]

# Reshape the DataFrame
# We will melt the DataFrame, keeping the id_vars and melting the question_cols
df_long = pd.melt(df, id_vars=id_vars, value_vars=question_cols, var_name='Question_Metric', value_name='Value')

# Separate the question number and the metric (Options, Key, Mark)
df_long[['Question', 'Metric']] = df_long['Question_Metric'].str.extract(r'(Q \d+) (.+)')

# Pivot the data to have 'Options', 'Key', and 'Mark' as columns for each question and MATRÍCULA
df_reshaped = df_long.pivot_table(index=id_vars + ['Question'], columns='Metric', values='Value', aggfunc='first').reset_index()

# Rename columns for clarity
df_reshaped.columns.name = None
df_reshaped = df_reshaped.rename(columns={'Options': 'Q_Options', 'Key': 'Q_Key', 'Marks': 'Q_Mark'})

# Display the reshaped DataFrame
display(df_reshaped.head())
```
