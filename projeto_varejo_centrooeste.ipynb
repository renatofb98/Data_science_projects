```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import sqlite3
import requests

conn = sqlite3.connect('/content/drive/MyDrive/projeto_varejo_centrooeste/varejo_co.db')

df_pmc = pd.read_sql("SELECT * FROM pmc_varejo", conn)
print(df_pmc.shape)
```

```python
municipios_pib = set(df_pib["codigo_ibge"].unique())
municipios_pop = set(df_pop["codigo_ibge"].unique())

print("só na população, não no PIB:", municipios_pop - municipios_pib)
print("só no PIB, não na população:", municipios_pib - municipios_pop)
```

```python
df_pop_raw[df_pop_raw["D1C"].str[:-1] == "510183"][["D1C", "D1N"]].drop_duplicates()
```

```python
df_pop = df_pop[df_pop["codigo_ibge"] != "510183"].copy()
print(df_pop["codigo_ibge"].nunique())  # deve dar 467 agora, batendo com o PIB
```

```python
df_pib_per_capita = df_pib.merge(df_pop, on=["codigo_ibge", "ano"], how="inner")

# PIB vem em MIL reais -- multiplica por 1000 antes de dividir pela população
df_pib_per_capita["pib_per_capita"] = (df_pib_per_capita["pib_mil_reais"] * 1000) / df_pib_per_capita["populacao"]

print(df_pib_per_capita.shape)
df_pib_per_capita[df_pib_per_capita["codigo_ibge"] == "520870"][["ano", "pib_per_capita"]]  # confere Goiânia como sanity check
```

```python
# pega o PIB per capita do ano mais recente disponível por município
df_pib_recente = df_pib_per_capita.sort_values("ano").groupby("codigo_ibge").tail(1)[
    ["codigo_ibge", "ano", "pib_per_capita"]
].rename(columns={"ano": "ano_pib"})

print(df_pib_recente["ano_pib"].value_counts())
df_pib_recente.head()
```

```python
ultimo_mes = df_pmc_recente["anomes"].max()

# .unique() antes do sort -- sem isso, cada mês repetido (1x por UF) distorce a contagem posicional
meses_unicos = sorted(df_pmc_recente[df_pmc_recente["anomes"] < ultimo_mes]["anomes"].unique())
mes_12m_atras = meses_unicos[-12]

print("mês mais recente:", ultimo_mes, "| referência 12m atrás (corrigido):", mes_12m_atras)

df_tendencia = df_pmc_recente[df_pmc_recente["anomes"].isin([ultimo_mes, mes_12m_atras])].pivot(
    index="uf", columns="anomes", values="indice_volume_sazonal"
)
df_tendencia.columns = ["indice_12m_atras", "indice_atual"]
df_tendencia["variacao_12m_pct"] = (df_tendencia["indice_atual"] / df_tendencia["indice_12m_atras"] - 1) * 100

df_tendencia.reset_index()
```

```python
df_municipios = df_pib_recente.merge(
    df_pop[df_pop["ano"] == df_pop["ano"].max()][["codigo_ibge", "populacao"]],
    on="codigo_ibge", how="left"
)

# traz a UF de cada município a partir do próprio código (os 2 primeiros dígitos do código de 6 já indicam a UF)
mapa_uf = {"50": "Mato Grosso do Sul", "51": "Mato Grosso", "52": "Goiás", "53": "Distrito Federal"}
df_municipios["uf"] = df_municipios["codigo_ibge"].str[:2].map(mapa_uf)

df_municipios = df_municipios.merge(df_tendencia.reset_index()[["uf", "variacao_12m_pct"]], on="uf", how="left")

print(df_municipios.shape)
df_municipios.isnull().sum()
```

```python
df_municipios["tercil_renda"] = pd.qcut(df_municipios["pib_per_capita"], 3, labels=["baixa", "média", "alta"])
df_municipios["tercil_populacao"] = pd.qcut(df_municipios["populacao"], 3, labels=["baixa", "média", "alta"])

def classificar_formato(row):
    renda = row["tercil_renda"]
    pop = row["tercil_populacao"]

    if renda == "baixa" and pop == "baixa":
        return "Risco (sem escala nem renda)"
    elif renda == "baixa" and pop in ["média", "alta"]:
        return "Atacarejo"
    elif renda == "alta" and pop in ["média", "alta"]:
        return "Varejo tradicional"
    elif renda == "alta" and pop == "baixa":
        return "Varejo de nicho (renda alta, população pequena)"
    else:  # renda média
        return "Misto / avaliar caso a caso"

df_municipios["perfil_formato"] = df_municipios.apply(classificar_formato, axis=1)

print(df_municipios["perfil_formato"].value_counts())
df_municipios.groupby("perfil_formato")[["pib_per_capita", "populacao"]].mean()
```
